# Limpeza e Tratamento de Dados (Data Prep)

Agora que entendi o comportamento geral dos dados, vou preparar a minha base para os algoritmos. O meu primeiro passo é remover as linhas duplicadas, pois dados repetidos podem viciar o modelo. Em seguida, vou mapear a presença de dados nulos (ausentes) para decidir a melhor estratégia de preenchimento.

```python
# 1. Quantidade de linhas antes da limpeza
linhas_antes = dados.shape[0]

# Removendo as linhas exatas que estão duplicadas
dados = dados.drop_duplicates()

# Contando quantas linhas sobraram
linhas_depois = dados.shape[0]

print(f"Linhas antes: {linhas_antes}")
print(f"Linhas depois: {linhas_depois}")
print(f"Total de duplicadas removidas: {linhas_antes - linhas_depois}\n")

# 2. Procurar por 'buracos' na tabela (dados nulos/ausentes)
print("Quantidade de buracos (dados nulos) por coluna:")
display(dados.isnull().sum())
```

### Imputação de Dados e Análise de Outliers

Ao verificar a presença de dados nulos, identifiquei ausência de informação em 5% dos registos nas colunas de temperatura, rotação e torque.

**Justificativa de Imputação:** Optei por preencher (imputar) estes valores nulos utilizando a **Mediana** em vez da Média. A mediana é uma medida estatística robusta que não sofre grande influência caso existam valores extremos (outliers) na distribuição dos dados dos sensores da fábrica.

Abaixo, realizo este preenchimento numérico e, em seguida, gero gráficos de Boxplot para identificar visualmente a presença de outliers nas variáveis explicativas.

```python
# 1. Separando apenas as colunas com números para fazer o cálculo
colunas_numericas = dados.select_dtypes(include=['float64', 'int64']).columns

# 2. Tapando os buracos de todas as colunas numéricas usando a Mediana
dados[colunas_numericas] = dados[colunas_numericas].fillna(dados[colunas_numericas].median())

# 3. Verificando se os buracos sumiram
print("Quantidade de buracos após o tratamento:")
display(dados.isnull().sum())

# 4. Criando os gráficos Boxplot para procurar Outliers
import matplotlib.pyplot as plt
import seaborn as sns

# Preparando o espaço para os gráficos
plt.figure(figsize=(12, 5))

# Desenhando o Boxplot da Velocidade de Rotação
plt.subplot(1, 2, 1)
sns.boxplot(y=dados['velocidade_rotacao_rpm'], color='orange')
plt.title('Identificação de Outliers - Velocidade de Rotação')

# Desenhando o Boxplot do Torque
plt.subplot(1, 2, 2)
sns.boxplot(y=dados['torque_nm'], color='green')
plt.title('Identificação de Outliers - Torque')

# Mostrando na ecrã
plt.tight_layout()
plt.show()
```

# Feature Engineering

Para ajudar o algoritmo a encontrar padrões mais complexos, vou criar uma nova variável preditora. O documento da engenharia sugere que a potência aplicada pelas máquinas pode ser um forte indicativo de estresse mecânico.

Portanto, vou criar a coluna `potencia`, resultante da multiplicação matemática entre a `velocidade_rotacao_rpm` e o `torque_nm`. Como já realizei o tratamento de nulos na etapa anterior, espero que esta operação não ocorra erros ou valores vazios.

```python
# Criando a nova coluna 'potencia' através da multiplicação das duas variáveis
dados['potencia'] = dados['velocidade_rotacao_rpm'] * dados['torque_nm']

# Exibindo as primeiras linhas para confirmar que a nova coluna foi adicionada no final da tabela
display(dados.head())
```

# Divisão e Balanceamento dos Dados

Nesta etapa, vou preparar os dados para o treinamento do modelo.

Primeiro, separando a variável alvo `falha_maquina` (y) das variáveis preditoras (X). Para evitar problema de **Data Leakage (Vazamento de Dados)**, estou removendo intencionalmente as colunas de falhas específicas (`falha_twf`, `falha_hdf`, `falha_pwf`, `falha_osf`, `falha_rnf`), além das colunas de identificação e texto que não agregam valor preditivo (`udi`, `id_produto`, `tipo`).

Em seguida, os dados serão divididos em 80% para treino e 20% para teste, utilizando o parâmetro `stratify=y` para garantir que a proporção de falhas se mantenha em ambos os conjuntos. Por fim, aplico a técnica de reamostragem **SMOTE** exclusivamente nos dados de treino. Isso previne que o modelo sofra overfitting na classe majoritária, sem contaminar os dados de teste com amostras sintéticas.

```python
# Trazendo as ferramentas do scikit-learn (a principal biblioteca de Machine Learning)
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE

# 1. Separando as variáveis (X = características da máquina, y = o alarme de falha)
# Vou apagar do 'X' as colunas de identificação, a resposta final e as colunas que não agregam valor preditivo.
X = dados.drop(columns=['falha_maquina', 'falha_twf', 'falha_hdf', 'falha_pwf', 'falha_osf', 'falha_rnf', 'udi', 'id_produto', 'tipo'])
y = dados['falha_maquina']

# 2. Dividindo os dados (80% para treino, 20% para teste)
# O stratify=y garante que a mesma porcentagem de falhas vá para o treino e para o teste
X_treino, X_teste, y_treino, y_teste = train_test_split(X, y, test_size=0.20, stratify=y, random_state=42)

# 3. Aplicando o SMOTE apenas nos dados de treino para balancear
smote = SMOTE(random_state=42)
X_treino_balanceado, y_treino_balanceado = smote.fit_resample(X_treino, y_treino)

# Verificando como ficou o balanceamento
print("Contagem de falhas ANTES do SMOTE (dados originais de treino):")
display(y_treino.value_counts())

print("\nContagem de falhas DEPOIS do SMOTE (dados de treino balanceados):")
display(y_treino_balanceado.value_counts())
```

# Escalonamento de Variáveis (StandardScaler)

Nesta etapa, preparo os dados especificamente para os dois modelos que serão avaliados.

Vou aplicar o `StandardScaler` para padronizar as escalas das variáveis numéricas, mas **exclusivamente para o modelo KNN**. O KNN funciona calculando a distância euclidiana entre os pontos de dados; portanto, variáveis com grandezas maiores (como a rotação em milhares) ofuscariam o peso de variáveis menores (como o torque). Aplico o método `fit_transform` nos dados de treino e apenas `transform` nos dados de teste para impedir o vazamento de dados.

**Justificativa para a Árvore de Decisão:** Os dados destinados ao modelo de Árvore de Decisão (Decision Tree) serão mantidos em sua escala original. Modelos baseados em árvores funcionam através de cortes (splits) sequenciais baseados em regras lógicas (ex: temperatura > 300) e não calculam distâncias entre os pontos. Por isso, são matematicamente imunes e insensíveis à magnitude ou escala dos atributos.

```python
# Ferramenta para padronizar a "régua" dos dados
from sklearn.preprocessing import StandardScaler

# 1. Escalonador
scaler = StandardScaler()

# 2. Criando cópias dos dados balanceados especificamente para o KNN
X_treino_knn = X_treino_balanceado.copy()
X_teste_knn = X_teste.copy()

# 3. Aplicando a régua (StandardScaler) APENAS nos dados do KNN
# fit_transform no treino: o modelo "aprende" a régua com os dados de estudo e já a aplica
X_treino_knn = scaler.fit_transform(X_treino_knn)

# transform no teste: o modelo aplica a mesma régua aprendida, sem "espiar" as respostas do teste
X_teste_knn = scaler.transform(X_teste_knn)

# (para a Árvore de Decisão vou usar os originais: X_treino_balanceado e X_teste)

print("Escalonamento para o modelo KNN realizado com sucesso!")
print("Dados para a Árvore de Decisão preservados em sua escala original.")
```

# Ajuste de Parâmetros e Combate ao Overfitting

Treinamento dos dois modelos (KNN e Árvore de Decisão) e ajuste dos  hiperparâmetros para encontrar o ponto de equilíbrio ideal, evitando o overfitting (sobreajuste).

*   Para o **KNN**, vou testar o número de vizinhos (`n_neighbors`) com os valores 3, 5 e 7.
*   Para a **Árvore de Decisão**, vou limitar a profundidade máxima (`max_depth`) testando os valores 3, 5 e `None` (crescimento total).

O objetivo é registar a acurácia (taxa de acerto) tanto nos dados de treino quanto nos dados de teste para identificar onde ocorre a perda de generalização (o momento exato em que o modelo decora os dados de treino, mas falha no teste).

```python
# Trazendo os "cérebros" da Inteligência Artificial e a ferramenta de dar notas (Acurácia)
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# --- TREINANDO O MODELO KNN ---
print("--- RESULTADOS DO KNN ---")
vizinhos = [3, 5, 7]
for k in vizinhos:
    # Criando o modelo com o parâmetro atual
    knn = KNeighborsClassifier(n_neighbors=k)
    # Treinando o modelo (usando os dados escalonados!)
    knn.fit(X_treino_knn, y_treino_balanceado)

    # Calculando a nota no treino e no teste
    nota_treino_knn = accuracy_score(y_treino_balanceado, knn.predict(X_treino_knn))
    nota_teste_knn = accuracy_score(y_teste, knn.predict(X_teste_knn))

    print(f"KNN (K={k}): Nota no Treino = {nota_treino_knn:.4f} | Nota no Teste = {nota_teste_knn:.4f}")

# --- TREINANDO A ÁRVORE DE DECISÃO ---
print("\n--- RESULTADOS DA ÁRVORE DE DECISÃO ---")
profundidades = [3, 5, None]
for prof in profundidades:
    # Criando o modelo com o parâmetro atual
    arvore = DecisionTreeClassifier(max_depth=prof, random_state=42)
    # Treinando o modelo (usando os dados normais, sem escala!)
    arvore.fit(X_treino_balanceado, y_treino_balanceado)

    # Calculando a nota no treino e no teste
    nota_treino_arvore = accuracy_score(y_treino_balanceado, arvore.predict(X_treino_balanceado))
    nota_teste_arvore = accuracy_score(y_teste, arvore.predict(X_teste))

    print(f"Árvore (max_depth={prof}): Nota no Treino = {nota_treino_arvore:.4f} | Nota no Teste = {nota_teste_arvore:.4f}")
```

# Avaliação da Acurácia e Conclusão

### Análise de Overfitting e Desempenho
*   **KNN:** O modelo apresentou o seu melhor desempenho com o parâmetro `n_neighbors=3` (90,65% no teste), perdendo precisão nas configurações subsequentes.
*   **Árvore de Decisão:** Identifiquei a ocorrência clara de *overfitting* quando o parâmetro `max_depth=None` foi utilizado. O modelo sobreajustou-se aos dados de treino (atingindo 98,96%), o que indica memorização. A configuração com `max_depth=5` provou ser a mais estável e imune ao *overfitting* (Treino: 89,84% | Teste: 90,65%).

### Conclusão e Recomendação
Comparando as taxas de acerto na base de teste (dados inéditos), a **Árvore de Decisão** apresentou o desempenho superior. Mesmo com o comportamento de sobreajuste na sua versão sem limites de profundidade (`None`), ela alcançou a maior acurácia final (93,85%).

Portanto, o modelo recomendado para adoção pelo parque fabril é a **Árvore de Decisão**, por lidar melhor com as relações complexas das variáveis e dispensar o escalonamento, facilitando a implementação em produção.
